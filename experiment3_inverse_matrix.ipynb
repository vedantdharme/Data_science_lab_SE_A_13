```python
import numpy as np
```

```python
# Define a sample matrix
matrix_a = np.array([
    [1, 2],
    [3, 4]
])

print("Original Matrix A:")
print(matrix_a)
```

```python
# Calculate the inverse of the matrix
try:
    inverse_matrix_a = np.linalg.inv(matrix_a)
    print("\nInverse of Matrix A:")
    print(inverse_matrix_a)
except np.linalg.LinAlgError:
    print("\nError: The matrix is singular and does not have an inverse.")
```

You can replace `matrix_a` with your own matrix. Please ensure it's a square matrix and non-singular (its determinant is not zero) for the inverse to exist.
