### Minimax Algorithm

Minimax is a decision-making algorithm used in artificial intelligence, game theory, and other fields. It's primarily used for two-player, zero-sum games (where one player's gain is the other player's loss) with perfect information (both players know the full state of the game).

The algorithm works by recursively exploring the game tree, assigning a score to each possible game state. It assumes that both players play optimally: the 'maximizing' player tries to maximize their score, while the 'minimizing' player tries to minimize the opponent's score (which, from their perspective, maximizes their own).

**Key components of Minimax:**

1.  **Game State:** A representation of the current situation in the game.
2.  **Terminal State:** A game state where the game has ended (win, loss, or draw).
3.  **Evaluation Function:** A function that assigns a numerical score to a terminal state (e.g., +1 for a win, -1 for a loss, 0 for a draw). For non-terminal states, a heuristic can be used if the search depth is limited.
4.  **Maximizing Player:** The player trying to achieve the highest possible score.
5.  **Minimizing Player:** The player trying to achieve the lowest possible score (which means making the maximizing player's score as low as possible).

**How it works (simplified):**

*   **Maximize player's turn:** They choose the move that leads to the highest score from the next possible states.
*   **Minimize player's turn:** They choose the move that leads to the lowest score from the next possible states.

The recursion continues until a terminal state is reached or a predefined depth limit is met. The scores are then propagated back up the game tree.

```python
class GameState:
    """A simple example game state: players take turns adding 1, 2, or 3 to a total. The first player to reach 10 wins."""
    def __init__(self, current_total=0, current_player=1): # Player 1 is maximizing, Player -1 is minimizing
        self.current_total = current_total
        self.current_player = current_player
        self.target_score = 10

    def is_terminal(self):
        return self.current_total >= self.target_score

    def get_winner(self):
        if self.current_total >= self.target_score:
            return self.current_player * -1 # The player who made the last move wins, so current_player * -1 is the winner
        return 0 # No winner yet

    def get_possible_moves(self):
        if self.is_terminal():
            return []
        return [1, 2, 3]

    def apply_move(self, move):
        new_total = self.current_total + move
        new_player = self.current_player * -1 # Switch player
        return GameState(new_total, new_player)

    def evaluate(self):
        if self.current_total >= self.target_score:
            if self.current_player == 1: # If it's P1's turn, but P2 just reached target, P2 wins
                return -1
            else: # If it's P2's turn, but P1 just reached target, P1 wins
                return 1
        return 0 # Should not happen in a terminal state call, but for completeness

    def __str__(self):
        return f"Total: {self.current_total}, Current Player: {'Max' if self.current_player == 1 else 'Min'}"
```

```python
def minimax(game_state, depth, is_maximizing_player):
    """Implements the Minimax algorithm."""

    if depth == 0 or game_state.is_terminal():
        return game_state.evaluate()

    if is_maximizing_player:
        max_eval = -float('inf')
        for move in game_state.get_possible_moves():
            new_state = game_state.apply_move(move)
            eval = minimax(new_state, depth - 1, False)
            max_eval = max(max_eval, eval)
        return max_eval
    else:
        min_eval = float('inf')
        for move in game_state.get_possible_moves():
            new_state = game_state.apply_move(move)
            eval = minimax(new_state, depth - 1, True)
            min_eval = min(min_eval, eval)
        return min_eval
```

### Finding the Best Move

To find the best move, we iterate through all possible moves from the current state and apply the minimax algorithm to each resulting state. The move that leads to the best outcome for the current player is chosen.

```python
def find_best_move(game_state, depth):
    best_move = None
    if game_state.current_player == 1: # Maximizing player
        best_eval = -float('inf')
        for move in game_state.get_possible_moves():
            new_state = game_state.apply_move(move)
            eval = minimax(new_state, depth - 1, False)
            if eval > best_eval:
                best_eval = eval
                best_move = move
    else: # Minimizing player
        best_eval = float('inf')
        for move in game_state.get_possible_moves():
            new_state = game_state.apply_move(move)
            eval = minimax(new_state, depth - 1, True)
            if eval < best_eval:
                best_eval = eval
                best_move = move
    return best_move, best_eval

# --- Demonstration ---
print("--- Starting Game Simulation ---")
current_game = GameState(current_total=0, current_player=1) # Max player starts
max_depth = 10 # Adjust depth as needed for game complexity

while not current_game.is_terminal():
    print(f"\n{current_game}")
    move, score = find_best_move(current_game, max_depth)
    print(f"AI (Player {'Max' if current_game.current_player == 1 else 'Min'}) chooses move: {move} (Expected score: {score})")
    current_game = current_game.apply_move(move)

print(f"\nGame Over! Final {current_game}")
winner = current_game.get_winner()
if winner == 1:
    print("Player 1 (Max) wins!")
elif winner == -1:
    print("Player -1 (Min) wins!")
else:
    print("It's a draw!") # Not possible in this specific game, but good practice.
```
