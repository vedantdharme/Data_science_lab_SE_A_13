```python
import numpy as np

# Simulate a coin toss experiment
num_tosses = 1000 # Number of coin tosses
results = np.random.choice(['Heads', 'Tails'], size=num_tosses)

# Compute probabilities
num_heads = np.sum(results == 'Heads')
num_tails = np.sum(results == 'Tails')

probability_heads = num_heads / num_tosses
probability_tails = num_tails / num_tosses

print(f"Total number of tosses: {num_tosses}")
print(f"Number of Heads: {num_heads}")
print(f"Number of Tails: {num_tails}")
print(f"Probability of Heads: {probability_heads:.4f}")
print(f"Probability of Tails: {probability_tails:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Generate a normal distribution
mu, sigma = 0, 0.1 # mean and standard deviation
s = np.random.normal(mu, sigma, 1000)

# Plot its histogram
count, bins, ignored = plt.hist(s, 30, density=True)
plt.plot(bins, 1/(sigma * np.sqrt(2 * np.pi)) * np.exp( - (bins - mu)**2 / (2 * sigma**2) ),
         linewidth=2, color='r')
plt.title('Histogram of Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.show()
```
