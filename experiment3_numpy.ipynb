```python
import numpy as np

# Define the coefficient matrix A
A = np.array([[2, 3], [4, 5]])

# Define the constant vector B
B = np.array([10, 20])

# Solve the system of equations
X = np.linalg.solve(A, B)

# Print the solution
print(f"The solution to the system of equations is: x = {X[0]}, y = {X[1]}")
```
