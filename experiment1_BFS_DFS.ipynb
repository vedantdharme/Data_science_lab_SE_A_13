### 1. Representing a Graph

We'll use an adjacency list to represent our graph. Each key in the dictionary will represent a node, and its value will be a list of its neighbors.

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print("Sample Graph:")
for node, neighbors in graph.items():
    print(f"{node}: {neighbors}")
```

### 2. Breadth-First Search (BFS)

BFS explores all the neighbor nodes at the present depth prior to moving on to the nodes at the next depth level. It uses a queue data structure to keep track of nodes to visit.

**Algorithm Steps:**
1.  Start by putting any one of the graph's vertex at the back of a queue.
2.  Take the front item of the queue and add it to the visited list.
3.  Add all the neighbors of the dequeued vertex to the back of the queue, if they have not been visited yet.
4.  Repeat steps 2 and 3 until the queue is empty.

```python
from collections import deque

def bfs(graph, start_node):
    visited = [] # To keep track of visited nodes
    queue = deque([start_node]) # Initialize a queue with the start node

    while queue:
        node = queue.popleft() # Dequeue a node
        if node not in visited:
            visited.append(node) # Mark it as visited
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor) # Enqueue unvisited neighbors
    return visited

print("BFS Traversal starting from node 'A':")
bfs_result = bfs(graph, 'A')
print(bfs_result)
```

### 3. Depth-First Search (DFS)

DFS explores as far as possible along each branch before backtracking. It uses a stack data structure (or recursion, which uses the call stack) to keep track of nodes to visit.

**Algorithm Steps (Iterative using a stack):**
1.  Push the starting node onto the stack.
2.  While the stack is not empty:
    a. Pop a node from the stack.
    b. If the node has not been visited, mark it as visited and add it to the traversal list.
    c. Push all its unvisited neighbors onto the stack (order might matter, often reversed for consistent output if using set/dict for neighbors).
4.  Repeat until the stack is empty.

```python
def dfs(graph, start_node):
    visited = [] # To keep track of visited nodes
    stack = [start_node] # Initialize a stack with the start node

    while stack:
        node = stack.pop() # Pop a node from the stack
        if node not in visited:
            visited.append(node) # Mark it as visited
            # Push unvisited neighbors onto the stack
            # For consistent output, reverse the neighbors if they are sorted
            # to process them in the order they appear in the graph (e.g., C before B).
            for neighbor in reversed(graph[node]): # Reverse to get C before B if graph[A] is ['B', 'C']
                if neighbor not in visited:
                    stack.append(neighbor)
    return visited

print("DFS Traversal starting from node 'A':")
dfs_result = dfs(graph, 'A')
print(dfs_result)
```
